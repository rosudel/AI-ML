```python
## Pdf reader
from langchain_community.document_loaders import PyPDFLoader
# loader=PyPDFLoader('attention.pdf')
loader=PyPDFLoader('Younus.pdf')
docs=loader.load()
```

```python
docs
```

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter
text_splitter=RecursiveCharacterTextSplitter(chunk_size=1000,chunk_overlap=200)
documents=text_splitter.split_documents(docs)
documents[:5]
```

```python
from langchain_ollama import OllamaEmbeddings
from langchain.vectorstores import Chroma

# Initialize the Ollama Embeddings with the specified model
embeddings = OllamaEmbeddings(
    model="llama3.2:1b",  # Specify the model name
)

# Use the initialized embeddings in the Chroma vector store(database)

db = Chroma.from_documents(documents, embeddings)

# import sqlite3
# print(sqlite3.sqlite_version)  #its check sqlite3 version, required version is 3.35.0 or more to run chromadb
```

```python
#Vector database
query = "Yunus joined the Bureau of Economics as a researc"
result = db.similarity_search(query)
result
# result[0].page_content
```
